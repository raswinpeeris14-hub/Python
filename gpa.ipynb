```python
while True:
  print("==========Grading System=========")
  tam=int(input("\n Enter The mark for tamil"))
  eng=int(input("\nEnter The mark for english"))
  mat=int(input("\nEnter The mark for maths"))
  sci=int(input("\nEnter The mark for science"))
  soc=int(input("\nEnter The mark for social"))

  count=500
  total_value=tam+eng+mat+sci+soc
  avg_grade=(total_value/500)*100
  print(avg_grade)

  if avg_grade>=90:
    print("O")
  elif avg_grade>=75 and avg_grade<=90:
    print("A")
  elif avg_grade>=65 and avg_grade<=75:
    print("B")
  elif avg_grade>=55 and avg_grade<=65:
    print("C")
  elif avg_grade>=45 and avg_grade<=55:
    print("D")
  else:
    print("You are failed!")

  print("Wish To continue:Y/N:")
  choice=input("Enter the choice")
  if choice == "No":
    break
  else:
    continue
```
