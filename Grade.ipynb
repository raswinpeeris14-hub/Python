```python
tamil_mark=int(input("Tamil:"))
english_mark=int(input("English:"))
maths_mark=int(input("Maths:"))
physics_mark=int(input("Physics:"))
chemistry_mark=int(input("Chemistry:"))
computer_mark=int(input("Computer Science:"))

total_mark=tamil_mark+english_mark+maths_mark+physics_mark+chemistry_mark+computer_mark
total_value=600
print(f"\nTotal Marks:{total_mark}")

average_score=int((total_mark/total_value)*100)
print(f"\naverage_score:{average_score}")

if average_score >=90:
    print(f"60% discount for Schloarship in our college for {total_mark}")
elif average_score >=80 and average_score <=89:
    print(f"50% discount for Schloarship in our college for {total_mark}")
elif average_score >=70 and average_score <=79:
    print(f"40% discount for Schloarship in our college for {total_mark}")
elif average_score >=60 and average_score <=69:
    print(f"30% discount for Schloarship in our college for {total_mark}")
elif average_score >=50 and average_score <=59:
    print(f"20% discount for Schloarship in our college for {total_mark}")
elif average_score >=40 and average_score <=49:
    print(f"10% discount for Schloarship in our college for {total_mark}")
else:
    print(f"You need to pay Full Amount and No schloarship is available for {total_mark}")
```
