```
print("=============Greater Values=============")
num1=int(input("Enter The Number 1:"))
num2=int(input("\nEnter The Number 2:"))
num3=int(input("\nEnter The Number 3:"))
num4=int(input("\nEnter The Number 4:"))
print("\n")

if num1 > num2 and num1 > num3 and num1 > num4:
  print(f"{num1} is greater than {num2},{num3},{num4}")
elif num2 > num1 and num2 > num3 and num2 > num4:
  print(f"{num2} is greater than {num1},{num3},{num4}")
elif num3 > num1 and num3 > num2 and num2 > num4:
  print(f"{num3} is greater than {num1},{num2},{num4}")
elif num4 > num1 and num4 > num2 and num4 > num3:
  print(f"{num4} is greater than {num1},{num2},{num3}")
else:
  print("All numbers Are lower and not greater")
```

```
print("=============Sum of 1 to 100=============")
start_value=1
end_value=100
total=0
for i in range(start_value,end_value+1):
  total+=i
print(f"sum of numbers from {start_value} to {end_value} is:{total}")
```

```
print("=============Reverse a string=============")
name=input("Enter the name->")
reverse=" "
for i in name:
  reverse=i+reverse
print(f"Reverse String is:{reverse}")
```

```
print("=============Count Vowels in a String=============")
input_String=input("Enter a string value->")
print("\n")
vowels="aeiouAEIOU"
count_value=0
vowels_found=" "
for a in input_String:
  for b in vowels:
    if a == b:
      count_value+=1
      vowels_found+=a
print(f"The Vowels in the string are:{count_value}")
print(f"The Vowels are {vowels_found}")
```
